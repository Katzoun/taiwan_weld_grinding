```python
import math

# 四元數轉由拉角
def quaternion_to_euler(w, x, y, z):
    # 計算俯仰角 (Pitch)
    sin_theta = 2 * (w * y - z * x)
    if abs(sin_theta) >= 1:
        pitch = math.copysign(math.pi / 2, sin_theta)  # 避免數值超過範圍
    else:
        pitch = math.asin(sin_theta)
    
    # 計算滾轉角 (Roll)
    roll = math.atan2(2 * (w * x + y * z), 1 - 2 * (x**2 + y**2))
    
    # 計算偏航角 (Yaw)
    yaw = math.atan2(2 * (w * z + x * y), 1 - 2 * (y**2 + z**2))
    
    # 將弧度轉為角度
    roll_deg = math.degrees(roll)
    pitch_deg = math.degrees(pitch)
    yaw_deg = math.degrees(yaw)
    
    return roll_deg, pitch_deg, yaw_deg

# 由拉角轉四元數
def euler_to_quaternion(roll_deg, pitch_deg, yaw_deg):
    # 將角度轉為弧度
    roll = math.radians(roll_deg)
    pitch = math.radians(pitch_deg)
    yaw = math.radians(yaw_deg)
    
    # 計算四元數
    w = math.cos(roll / 2) * math.cos(pitch / 2) * math.cos(yaw / 2) + math.sin(roll / 2) * math.sin(pitch / 2) * math.sin(yaw / 2)
    x = math.sin(roll / 2) * math.cos(pitch / 2) * math.cos(yaw / 2) - math.cos(roll / 2) * math.sin(pitch / 2) * math.sin(yaw / 2)
    y = math.cos(roll / 2) * math.sin(pitch / 2) * math.cos(yaw / 2) + math.sin(roll / 2) * math.cos(pitch / 2) * math.sin(yaw / 2)
    z = math.cos(roll / 2) * math.cos(pitch / 2) * math.sin(yaw / 2) - math.sin(roll / 2) * math.sin(pitch / 2) * math.cos(yaw / 2)
    
    return w, x, y, z

# 主程式
if __name__ == "__main__":
    print("選擇功能:")
    print("1: 輸入四元數轉為由拉角")
    print("2: 輸入由拉角轉為四元數")
    choice = input("請選擇 (1 或 2): ")

    if choice == "1":
        # 輸入四元數
        w = float(input("輸入四元數 w: "))
        x = float(input("輸入四元數 x: "))
        y = float(input("輸入四元數 y: "))
        z = float(input("輸入四元數 z: "))
        
        # 計算由拉角
        roll, pitch, yaw = quaternion_to_euler(w, x, y, z)
        print(f"滾轉角: {roll:.2f}°")
        print(f"俯仰角: {pitch:.2f}°")
        print(f"偏航角: {yaw:.2f}°")

    elif choice == "2":
        # 輸入由拉角
        roll_deg = float(input("輸入滾轉角 (Roll-x) (°): "))
        pitch_deg = float(input("輸入俯仰角 (Pitch-y) (°): "))
        yaw_deg = float(input("輸入偏航角 (Yaw-z) (°): "))
        
        # 計算四元數
        w, x, y, z = euler_to_quaternion(roll_deg, pitch_deg, yaw_deg)
        print(f"四元數: w={w:.4f}, x={x:.4f}, y={y:.4f}, z={z:.4f}")
    else:
        print("無效選擇！")
```
