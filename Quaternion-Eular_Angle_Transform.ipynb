```python
import math

# Convert quaternion to Euler angles
def quaternion_to_euler(w, x, y, z):
    # Compute pitch
    sin_theta = 2 * (w * y - z * x)
    if abs(sin_theta) >= 1:
        pitch = math.copysign(math.pi / 2, sin_theta)  # clamp to avoid out-of-range values
    else:
        pitch = math.asin(sin_theta)
    
    # Compute roll
    roll = math.atan2(2 * (w * x + y * z), 1 - 2 * (x**2 + y**2))
    
    # Compute yaw
    yaw = math.atan2(2 * (w * z + x * y), 1 - 2 * (y**2 + z**2))
    
    # Convert radians to degrees
    roll_deg = math.degrees(roll)
    pitch_deg = math.degrees(pitch)
    yaw_deg = math.degrees(yaw)
    
    return roll_deg, pitch_deg, yaw_deg

# Convert Euler angles to quaternion
def euler_to_quaternion(roll_deg, pitch_deg, yaw_deg):
    # Convert degrees to radians
    roll = math.radians(roll_deg)
    pitch = math.radians(pitch_deg)
    yaw = math.radians(yaw_deg)
    
    # Compute quaternion
    w = math.cos(roll / 2) * math.cos(pitch / 2) * math.cos(yaw / 2) + math.sin(roll / 2) * math.sin(pitch / 2) * math.sin(yaw / 2)
    x = math.sin(roll / 2) * math.cos(pitch / 2) * math.cos(yaw / 2) - math.cos(roll / 2) * math.sin(pitch / 2) * math.sin(yaw / 2)
    y = math.cos(roll / 2) * math.sin(pitch / 2) * math.cos(yaw / 2) + math.sin(roll / 2) * math.cos(pitch / 2) * math.sin(yaw / 2)
    z = math.cos(roll / 2) * math.cos(pitch / 2) * math.sin(yaw / 2) - math.sin(roll / 2) * math.sin(pitch / 2) * math.cos(yaw / 2)
    
    return w, x, y, z

# Main program
if __name__ == "__main__":
    print("Choose a function:")
    print("1: Convert quaternion to Euler angles")
    print("2: Convert Euler angles to quaternion")
    choice = input("Choose (1 or 2): ")

    if choice == "1":
        # Input quaternion
        w = float(input("Enter quaternion w: "))
        x = float(input("Enter quaternion x: "))
        y = float(input("Enter quaternion y: "))
        z = float(input("Enter quaternion z: "))
        
        # Compute Euler angles
        roll, pitch, yaw = quaternion_to_euler(w, x, y, z)
        print(f"Roll: {roll:.2f}°")
        print(f"Pitch: {pitch:.2f}°")
        print(f"Yaw: {yaw:.2f}°")

    elif choice == "2":
        # Input Euler angles
        roll_deg = float(input("Enter roll (Roll-x) (°): "))
        pitch_deg = float(input("Enter pitch (Pitch-y) (°): "))
        yaw_deg = float(input("Enter yaw (Yaw-z) (°): "))
        
        # Compute quaternion
        w, x, y, z = euler_to_quaternion(roll_deg, pitch_deg, yaw_deg)
        print(f"Quaternion: w={w:.4f}, x={x:.4f}, y={y:.4f}, z={z:.4f}")
    else:
        print("Invalid choice!")
```
